```python
import numpy as np
import matplotlib.pyplot as plt

# Dataset
delays = [15, 30, 45, 20, 25, 100, 5, 60, 35, 50, 
          120, 80, 10, 75, 90, 200, 55, 40, 70, 150]

# Convert to numpy array
delays_arr = np.array(delays)

# Percentiles
percentiles = np.percentile(delays_arr, [10, 25, 50, 75, 90])
print("Percentiles (10th, 25th, 50th, 75th, 90th):", percentiles)

# IQR Calculation
Q1 = percentiles[1]
Q3 = percentiles[3]
IQR = Q3 - Q1
print("IQR:", IQR)

# Outlier Detection using IQR Rule
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = [x for x in delays if x < lower_bound or x > upper_bound]
print("Outliers:", outliers)

plt.figure(figsize=(5, 6))
plt.boxplot(delays)

plt.title("Box Plot of Flight Delays (in minutes)")
plt.ylabel("Delay Time (minutes)")

plt.show()
```

Question 2

```python
import numpy as np
import matplotlib.pyplot as plt
from statistics import mean, median, mode

# Dataset
salaries = [30000, 32000, 35000, 37000, 40000, 42000, 45000, 47000, 50000, 55000,
            60000, 62000, 65000, 67000, 70000, 72000, 75000, 80000, 85000, 90000]

# Mean
mean_salary = mean(salaries)
print("Mean Salary:", mean_salary)

# Median
median_salary = median(salaries)
print("Median Salary:", median_salary)

# Mode
try:
    mode_salary = mode(salaries)
except:
    mode_salary = "No unique mode"
print("Mode Salary:", mode_salary)

# Skewness check
if mean_salary < median_salary:
    print("The data is Left Skewed (Negative Skew)")
elif mean_salary > median_salary:
    print("The data is Right Skewed (Positive Skew)")
else:
    print("The data is Symmetric")

plt.figure(figsize=(7,5))
plt.hist(salaries, bins=8)
plt.title("Histogram of Employee Salaries")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Dataset
data = {
    'Student': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Mathematics': [85, 78, 92, 88, 70, 95, 60, 80, 90, 76],
    'Science': [80, 85, 88, 70, 75, 92, 55, 82, 89, 78],
    'English': [78, 74, 85, 80, 68, 90, 50, 77, 83, 72]
}

df = pd.DataFrame(data)

# 1. Descriptive Statistics
print("=== Descriptive Statistics ===")
print(df.describe())

# 2. Histogram of Scores
df[['Mathematics', 'Science', 'English']].hist(bins=5, figsize=(10,5), color='skyblue', edgecolor='black')
plt.suptitle("Distribution of Student Scores")
plt.show()

# 3. Correlation Analysis
corr = df[['Mathematics', 'Science', 'English']].corr()
print("\n=== Correlation Matrix ===")
print(corr)

# 4. Heatmap
plt.figure(figsize=(6,5))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap of Subjects")
plt.show()
```

```python
from scipy.stats import ttest_ind

# Dataset
medication_group = [110, 115, 108, 102, 107, 99, 111, 104, 109, 101]
placebo_group = [120, 125, 130, 122, 128, 119, 124, 127, 123, 126]

# 1. Perform Independent T-Test
t_stat, p_value = ttest_ind(medication_group, placebo_group)

# 2. Print Results
print("T-Statistic:", t_stat)
print("P-Value:", p_value)

# 3. Hypothesis Decision
alpha = 0.05  # significance level
if p_value < alpha:
    print("Reject the null hypothesis: The medication is effective")
else:
    print("Fail to reject the null hypothesis: No significant effect of medication")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Dataset
ratings = [8, 9, 7, 5, 6, 10, 9, 4, 7, 8, 
           6, 9, 10, 5, 8, 7, 6, 9, 10, 7]
```

```python
# Calculate descriptive statistics
mean_rating = np.mean(ratings)
median_rating = np.median(ratings)
mode_rating = stats.mode(ratings, keepdims=True).mode[0]
std_dev = np.std(ratings)

print(f"Mean Rating: {mean_rating}")
print(f"Median Rating: {median_rating}")
print(f"Mode Rating: {mode_rating}")
print(f"Standard Deviation: {std_dev}")

# Visualizing the distribution using Histogram
sns.histplot(ratings, kde=True)
plt.xlabel('Customer Satisfaction Rating')
plt.ylabel('Frequency')
plt.title('Distribution of Customer Satisfaction Ratings')
plt.show()
```
